```python
df = df.sort_values(['new_id', 'Месяц'])
```

```python
df.head()
```

```python
df['rto_lag_1'] = df.groupby('new_id')['РТО'].shift(1)
```

```python
df.head(12)
```

```python
df['rto_lag2'] = df.groupby('new_id')['РТО'].shift(2)
```

```python
df['rto_mean_lag6'] = df.groupby('new_id')['РТО'].shift(1).rolling(6).mean()
```

```python
df.head(12)
```

```python
df.drop(['lag_1'], axis = 1, inplace = True)
```

```python
df.head(12)
```

```python
df_model = df.dropna(subset = ['rto_lag_1', 'rto_lag2', 'rto_mean_lag6'])
```

```python
df_model.head(12)
```

```python
train_data = df_model[df_model['Месяц'] <= 9]
```

```python
test_data = df_model[df_model['Месяц'] == 10]
```

```python
test_data.head()
```

```python
df_model.columns
```

```python
df.info()
```

```python
from sklearn.preprocessing import OrdinalEncoder
import numpy as np
```

```python
oe = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=np.nan)
```

```python
categorical_features = ['Дата открытия, категориальный', 
                        'Торговая площадь, категориальный', 
                        'Населенный пункт', 
                        'Регион']
```

```python
oe.fit(train_data[categorical_features])
```

```python
X_train_cat = oe.transform(train_data[categorical_features])
```

```python
X_test_cat = oe.transform(test_data[categorical_features])
```

```python
numeric_features = ['Месяц', 'Численность населения', 'Количество домохозяйств', 
                    'Трафик пеший, в час', 'Трафик авто, в час', 
                    'Маркетплейсы, доставки, постаматы (100 м)', 
                    'Медицинские уч. и аптеки (300 м)', 'Школы (300 м)', 
                    'Остановки (300 м)', 'Продуктовые магазины (500 м)', 
                    'Пятерочки (500 м)', 'Количество касс', 
                    'Флаг алкогольной лицензии', 'rto_lag_1', 'rto_lag2', 
                    'rto_mean_lag6']
```

```python
X_train = np.hstack([train_data[numeric_features].values, X_train_cat])
X_test = np.hstack([test_data[numeric_features].values, X_test_cat])
```

```python
Y_train = train_data['РТО']
Y_test = test_data['РТО']
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
tree_model = DecisionTreeRegressor()
```

```python
tree_model.fit(X_train, Y_train)
```

```python
y_pred_tree = tree_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
mape_tree = mean_absolute_percentage_error(Y_test, y_pred_tree) * 100
```

```python
mape_tree
```

```python
df_november = df_model[df_model['Месяц'] == 10].copy()
```

```python
X_november_cat = oe.transform(df_november[categorical_features])
```

```python
X_november_num = df_november[numeric_features].values
```

```python
X_november = np.hstack([X_november_num, X_november_cat])
```

```python
y_november = tree_model.predict(X_november)
```

```python
submission = pd.DataFrame({'new_id': df_november['new_id'], 'rto': y_november})
```
